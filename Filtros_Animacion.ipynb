# Animación de Filtros Digitales
Haz clic en el botón de **Play** (▶️) a la izquierda de la celda de abajo para ver la animación.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.signal import butter, filtfilt, firwin
from IPython.display import HTML

# 1. Señal y filtros
fs = 1000
t = np.linspace(0, 3, 3 * fs, endpoint=False)
signal_clean = np.sin(2 * np.pi * 50 * t) + np.sin(2 * np.pi * 120 * t)

np.random.seed(0)
noise = 0.5 * np.random.normal(size=len(t))
signal_noisy = signal_clean + noise

nyq = 0.5 * fs
b_lp, a_lp = butter(4, 80.0 / nyq, btype='low')
b_hp = firwin(101, 100.0 / nyq, pass_zero=False, window='hamming')
b_bp, a_bp = butter(4, [40.0 / nyq, 60.0 / nyq], btype='band')

filtered_lp = filtfilt(b_lp, a_lp, signal_noisy)
filtered_hp = filtfilt(b_hp, [1.0], signal_noisy)
filtered_bp = filtfilt(b_bp, a_bp, signal_noisy)

# 2. Configurar la gráfica
fig, axs = plt.subplots(4, 1, figsize=(10, 8))
fig.patch.set_facecolor('#FFF5F8')
plt.subplots_adjust(hspace=0.7)

titles = ['Señal con Ruido Original', 'Filtro Pasa Bajos (Aísla 50 Hz)', 'Filtro Pasa Altos (Aísla 120 Hz)', 'Filtro Pasa Bandas (Señal Limpia)']
colors = ['gray', 'blue', 'red', 'green']
data = [signal_noisy, filtered_lp, filtered_hp, filtered_bp]
lines = []

for i in range(4):
    axs[i].set_ylim(-3, 3)
    axs[i].set_title(titles[i], fontweight='bold')
    axs[i].set_ylabel('Amplitud')
    if i == 3:
        axs[i].set_xlabel('Tiempo [s]')
        
    line, = axs[i].plot([], [], color=colors[i], lw=2)
    lines.append(line)

def update(frame):
    start = frame
    end = frame + 200
    x = t[start:end]
    for i in range(4):
        lines[i].set_data(x, data[i][start:end])
        axs[i].set_xlim(x[0], x[-1])
    return lines

# 3. Crear animación web
ani = FuncAnimation(fig, update, frames=range(0, len(t)-200, 20), interval=40, blit=False)

# 4. Mostrar el reproductor interactivo en la web
plt.close() 
HTML(ani.to_jshtml())
```
